```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
stop_words = set(stopwords.words('english'))
lammatizer = WordNetLemmatizer()
```

```python
import re
def preprocess_text(text):
    #text lower case
    text = str(text).lower()
    #remove punctuation 
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    #words tokanatiation
    words = text.split()
    #remove stopwords
    words = [word for word in words  if word not in stop_words]

    #lemmatizer
    words = [lammatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

```python
df['tags']=df['tags'].apply(preprocess_text)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(stop_words = 'english',max_features= 50000,ngram_range=(1,2))
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
indices = pd.Series(range(len(df)), index=df['primaryTitle']).drop_duplicates()

def recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    sim_idx = sim_score.argsort()[::-1][1:n+1]

    return df['primaryTitle'].iloc[sim_idx]
```

```python
recommend('Kate & Leopold')
```
